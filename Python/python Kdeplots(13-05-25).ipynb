```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from io import StringIO
from scipy.stats import gaussian_kde
```

```python
data = pd.read_excel("C:/Users/Chet/Documents/practice_data.xlsx")
```

```python
data
```

```python
plt.figure(figsize=(12, 6))

plt.subplot(1, 2, 1)
sns.kdeplot(data['Interpersonal Skills Rating'], fill=True, color='skyblue')
plt.title('Distribution of Interpersonal Skills Rating')


plt.subplot(1, 2, 2)
sns.kdeplot(data['Decision-Making Skills Rating'], fill=True, color='salmon')
plt.title('Distribution of Decision-Making Skills Rating')

plt.tight_layout()
plt.show()
```

```python

plt.figure(figsize=(12, 6))


plt.subplot(1, 2, 1)
sns.kdeplot(data['Sales Revenue Generated'], fill=True, color='skyblue')
plt.title('Distribution of Interpersonal Skills Rating')


plt.subplot(1, 2, 2)
sns.kdeplot(data['Sales Revenue Generated'], fill=True, color='salmon')
plt.title('Distribution of Decision-Making Skills Rating')

plt.tight_layout()
plt.show()
```

## What is the range of highest number of sales revenue generated?

```python
max_revenue = data['Sales Revenue Generated'].max()
min_revenue = data['Sales Revenue Generated'].min()

print(f"Max : ₹{max_revenue}")
print(f"Min : ₹{min_revenue}")
```
