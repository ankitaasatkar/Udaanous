```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data = pd.read_csv('C:/Users/Chet/Documents/sales.csv')  
data
```

```python
print(data.head())
print(data.info())
```

## Clean and convert dates

```python
data['OrderDate'] = pd.to_datetime(data['OrderDate'])
data['Month'] = data['OrderDate'].dt.to_period('M')
```

## Sales by month

```python
monthly_sales = data.groupby('Month')['Sales'].sum()
monthly_sales.plot(kind='bar', title='Monthly Sales', figsize=(10,5))
plt.ylabel('Sales')
plt.show()
```

## Sales by region

```python
# Step 6: Sales by region
sns.barplot(x='Region', y='Sales', data=data)
plt.title('Sales by Region')
plt.show()
```

## : Top 5 products

```python
top_products = data.groupby('ProductName')['Sales'].sum().nlargest(5)
top_products.plot(kind='barh', title='Top 5 Products')
plt.xlabel('Sales')
plt.show()
```
