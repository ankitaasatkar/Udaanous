## Creating Total Purchase Amount Variable

```python
data['Total_Amount'] = data['quantity'] * data['price']

print(data[['order_id', 'product_name', 'quantity', 'price', 'Total_Amount']])
```

## Creating Customer Lifetime Value Variable

```python
data['Customer_Lifespan_Months'] = pd.to_datetime(data['order_date'])

data['Customer_Lifespan_Months'] = data['order_date'].dt.strftime('%B')

data['Customer_Lifespan_Months'] = data['order_date'].dt.month

print(data[['product_name', 'order_date', 'Customer_Lifespan_Months']])
```

```python
data['CLV'] = (data['Order_Amount'] * data['Customer_Lifespan_Months'])
print(data[['product_name','CLV', 'Order_Amount', 'quantity','Customer_Lifespan_Months']])
```

## datetime

```python
data['order_date'] = pd.to_datetime(data['order_date'], format='%m/%d/%Y')

data['year'] = data['order_date'].dt.year
data['month'] = data['order_date'].dt.month
data['day'] = data['order_date'].dt.day

print(data[['order_id', 'order_date', 'year', 'month', 'day']])
```
