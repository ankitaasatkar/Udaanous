## Project: Customer Purchase Behavior Analysis

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data = pd.read_csv('C:/Users/Chet/Documents/customers.csv')  
data
```

## Initial Data Exploration

```python
print(data.head())
```

```python
# Check for missing values
print(data.isnull().sum())
```

```python
print(data.describe())
```

## Visualize Gender Distribution

```python
plt.figure(figsize=(8, 6))
sns.countplot(x='Gender', data=data)
plt.title('Gender Distribution')
plt.xlabel('Gender')
plt.ylabel('Count')
plt.show()
```

## Analyze Spending Score by Gender

```python
plt.figure(figsize=(8, 6))
sns.boxplot(x='Gender', y='SpendingScore', data=data)
plt.title('Spending Score by Gender')
plt.show()
```

## Relationship Between Age and Spending Score

```python
plt.figure(figsize=(9, 6))
sns.scatterplot(x='Age', y='SpendingScore', hue='Gender', data=data)
plt.title('Age vs Spending Score')
plt.show()
```

## Category-Wise Spending Analysis

```python
category_avg = data.groupby('PurchaseCategory')['SpendingScore'].mean().reset_index()
plt.figure(figsize=(9, 6))
sns.barplot(x='PurchaseCategory', y='SpendingScore', data=category_avg)
plt.title('Average Spending Score by Purchase Category')
plt.ylabel('Avg Spending Score')
plt.show()
```

## Optional Deeper Analysis

```python
plt.figure(figsize=(10, 6)) 
sns.scatterplot(x='AnnualIncome', y='SpendingScore', hue='Gender', data=data)
plt.title('Income vs Spending Score')
plt.show()
```

#### Pie Chart of Purchase Category Distribution

```python
category_counts = data['PurchaseCategory'].value_counts()
plt.figure(figsize=(8, 8)) 
category_counts.plot.pie(autopct='%1.1f%%', startangle=90)
plt.title('Purchase Category Distribution')
plt.ylabel('')
plt.show()
```

#### Heatmap of Correlation

```python
plt.figure(figsize=(8, 6))
sns.heatmap(data[['Age', 'AnnualIncome', 'SpendingScore']].corr(), annot=True, cmap='coolwarm') 
plt.title('Correlation Matrix')
plt.show()
```
