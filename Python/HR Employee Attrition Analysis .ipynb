# Project : HR Employee Attrition Analysis 

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data = pd.read_csv('C:/Users/Chet/Documents/HR-Employee_data.csv')  
data
```

```python
print(data.shape)
print(data.dtypes)
data.head()
```

# Attrition by Department

```python
plt.figure(figsize=(10,8))
sns.countplot(data=data, x='Department', hue='Attrition', palette='Set2')

plt.title('Attrition by Department')
plt.xlabel('Department')
plt.ylabel('Number of Employees')
plt.legend(title='Attrition')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

# Monthly Income Distribution


```python
plt.figure(figsize=(10,8))
sns.histplot(data=data, x='MonthlyIncome', hue='Attrition', kde=True, bins=30)
plt.title('Monthly Income Distribution by Attrition Status')
plt.show()
```

#  Job Satisfaction vs Attrition


```python
plt.figure(figsize=(10,8))
sns.countplot(data=data, x='JobSatisfaction', hue='Attrition')
plt.title('Job Satisfaction vs Attrition')
plt.xlabel('Job Satisfaction (1 = Low, 4 = High)')
plt.ylabel('Count')
plt.show()
```

# Bonus Deep-Dive (Optional)

#### 1 Attrition vs Years at Company

```python
plt.figure(figsize=(9,6))
sns.boxplot(data=data, x='Attrition', y='YearsAtCompany')
plt.title('Attrition vs Years at Company')
plt.show()
```

#### 2 Gender-Based Attrition


```python
plt.figure(figsize=(10,8))
sns.countplot(data=data, x='Gender', hue='Attrition')
plt.title('Attrition by Gender')
plt.show()
```

#### 3 Income vs Work-Life Balance


```python
plt.figure(figsize=(10,8))
sns.boxplot(data=data, x='WorkLifeBalance', y='MonthlyIncome', hue='Attrition')
plt.title('Income vs Work-Life Balance & Attrition')
plt.show()
```
